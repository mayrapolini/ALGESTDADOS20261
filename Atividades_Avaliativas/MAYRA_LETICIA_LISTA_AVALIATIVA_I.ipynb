```python
# 15-Verificador de Acesso por Horário
# Leia a hora atual (0-23) e o tipo de usuário usando match/case ("admin", "operador", "visitante").
# Valide a hora. Para cada tipo, use if aninhado com elif para verificar o turno (manhã: 6-11, tarde: 12-17, noite: 18-23, madrugada: 0-5).
# Regras: admin acessa sempre, operador apenas manhã e tarde, visitante apenas tarde.
# Exiba se o acesso é permitido ou negado e o turno atual.

hora = int(input("Digite a hora (0-23): "))
usuario = input("Digite o tipo de usuário (admin, operador, visitante): ").lower()

if hora < 0 or hora > 23:
    print("Hora inválida")
else:
    if 6 <= hora <= 11:
        turno = "manhã"
    elif 12 <= hora <= 17:
        turno = "tarde"
    elif 18 <= hora <= 23:
        turno = "noite"
    else:
        turno = "madrugada"

match usuario:
        case "admin":
                acesso = True
        case "operador":
            if turno == "manhã" or turno == "tarde":
                acesso = True
            else:
                acesso = False
        case "visitante":
            if turno == "tarde":
                acesso = True
            else:
                acesso = False
        case _:
            print("Usuário inválido")
            acesso = None

if acesso is not None:
        print(f"Acesso permitido. Turno atual: {turno}.")
else:
        print(f"Acesso negado. Turno atual: {turno}.")
```

```python
# 16 - Simulador de Bomba de Combustível
# Leia o tipo de combustível usando match/case ("gasolina", "etanol", "diesel") e a quantidade de litros.
# Valide que a quantidade é positiva.
# Cada combustível tem um preço por litro definido como variável no código.
# Com if aninhado, aplique 3% de desconto se o abastecimento for acima de 40 litros. Exiba o subtotal, o desconto (se houver) e o total a pagar.

gasolina = 6.75
etanol = 1.99
diesel = 6.20

tipo = input("Digite o combustível (gasolina, etanol, diesel): ").lower()
quantidade = float(input("Digite a quantidade de litros: "))


match tipo:
    case "gasolina":
        preco = gasolina
    case "etanol":
        preco = etanol
    case "diesel":
        preco = diesel
    case _:
        print("Combustível inválido!")
        preco = None

if preco is not None:
    if quantidade <= 0:
        print("Quantidade inválida!")
    else:
        subtotal = quantidade * preco
        if quantidade > 40:
            desconto = subtotal * 0.03
        else:
            desconto = 0
        total = subtotal - desconto

        print(f"Subtotal: R${(subtotal):.2f}\nDesconto: R${desconto:.2f}\nTotal: R${total:.2f}")
```

```python
# 17 - Classificador de Velocidade
# Leia a velocidade em km/h e o tipo de via usando match/case ("urbana" → limite 60, "rodovia" → limite 100, "autopista" → limite 120).
# Valide que a velocidade não é negativa.
# Com if aninhado com elif, verifique: dentro do limite, excesso leve (até 20% acima), excesso moderado (até 50% acima) ou excesso grave (acima de 50%).
# Exiba o limite da via, a velocidade e a situação.

velocidade = float(input("Digite a velocidade (km/h): "))
via = input("Digite o tipo de via: ").lower()

match via:
    case "urbana":
        limite = 60
    case "rodovia":
        limite = 100
    case "autopista":
        limite = 120
    case _:
        print("Via inválida.")
        limite = None

if limite is not None:
    if velocidade < 0:
        print("Velocidade inválida!")
    else:
        if velocidade <= limite:
            situacao = "Dentro do limite"
        elif velocidade <= limite * 1.2:
            situacao = "Excesso leve"
        elif velocidade <= limite * 1.5:
            situacao = "Excesso moderado"
        else:
            situacao = "Excesso grave"

        print(f"Via: {via} - Limite: {limite} km/h - Velocidade: {velocidade} km/h - Situação: {situacao}")
```

```python
# 19 - Detector de Intervalo com Tolerância
# Leia um valor alvo, uma tolerância e um valor medido, todos como números positivos. Valide as entradas.
# Calcule os limites inferior (alvo - tolerância) e superior (alvo + tolerância).
# Use condicionais para classificar o valor medido em: dentro da faixa, abaixo do limite ou acima do limite.
# Calcule e exiba o desvio em relação ao alvo e a classificação.


alvo=float(input("Digite um valor alvo: "))
tolerancia=float(input("Digite a tolerância: "))
medido=float(input("Digite o valor medido: "))

if alvo >= 0 and tolerancia >= 0 and medido >= 0:
    limite_inferior = alvo - tolerancia
    limite_superior = alvo + tolerancia
    if medido < limite_inferior:
        print(f'Alvo: {alvo} - Desvio: {alvo-medido} - Classificação: Abaixo do limite')
    elif limite_inferior <= medido <= limite_superior :
        print(f'Alvo: {alvo} - Desvio: {alvo-medido} - Classificação: Dentro da faixa')
    else:
        print(f'Alvo: {alvo} - Desvio: {medido-alvo} - Classificação: Acima do limite')
```

```python
# 21 - Calculadora de Energia Elétrica
# Leia o consumo em kWh do mês. Valide que não é negativo.
# Use elif para calcular: 0-100 kWh → R$0,50/kWh; 101-200 → R$0,75/kWh; 201-300 → R$1,00/kWh; acima → R$1,50/kWh.
# Com if aninhado, se consumo acima de 300 kWh, pergunte se o cliente optou pelo modo econômico (s/n) e aplique 10% de desconto se sim.
# Exiba o valor da conta com e sem desconto.


consumo = float(input("Digite o consumo mensal de energia em kWh: "))

if consumo >= 0:
    if 0 <= consumo <= 100:
        valor = consumo * 0.50
    elif 101 <= consumo <= 200:
        valor = consumo * 0.75
    elif 201 <= consumo <= 300:
        valor = consumo * 1.00
    else:
        valor = consumo * 1.50

    desconto = 0

    if consumo > 300:
        modo = input("Optou pelo modo econômico (s/n)? ").lower()
        if modo == "s":
            desconto = valor * 0.10

    valor_final = valor - desconto

    print(f"Consumo: {consumo} kWh\nValor sem desconto: R${valor:.2f}\nDesconto: R${desconto:.2f}\nValor final: R${valor_final:.2f}")
else:
    print("Valor inválido")
```

```python
# 22 - Verificador de Intervalo Múltiplo
# Leia quatro valores: um intervalo mínimo A, um máximo B (com A < B), um número N e um divisor D (positivo e não nulo).
# Valide todas as entradas. Verifique com if aninhado: 1ª camada — se N está no intervalo [A, B]; 2ª camada (dentro) — se N é divisível por D;
# exiba a combinação correta de cada situação em 4 mensagens possíveis: dentro do intervalo e divisível, dentro e não divisível, fora do intervalo e divisível, fora e não divisível.


A = int(input('Digite um intervalo mínimo: '))
B = int(input('Digite um intervalo máximo: '))
N = int(input('Digite um número: '))
D = int(input('Digite um divisor (positivo e não nulo): '))


if A >= B:
    print('Intervalo inválido: A deve ser menor que B.')
elif D <= 0:
    print('Divisor inválido: deve ser positivo e não nulo.')
else:
    if A <= N <= B:
        if N % D == 0:
            print(f'O número {N} está dentro do intervalo [{A},{B}] e é divisível por {D}')
        else:
            print(f'O número {N} está dentro do intervalo [{A},{B}] e não é divisível por {D}')
    else:
        if N % D == 0:
            print(f'O número {N} está fora do intervalo [{A},{B}] e é divisível por {D}')
        else:
            print(f'O número {N} está fora do intervalo [{A},{B}] e não é divisível por {D}')
```

```python
# 23 - Sistema de Pontuação de Jogo
# Leia a pontuação de três jogadores (0 a 1000 cada). Valide todas.
# Usando apenas variáveis e condicionais, determine: quem tem a maior pontuação (1º lugar), quem tem a menor (3º lugar) e quem é o do meio (2º lugar).
#Trate empates explicitamente. Calcule a diferença entre o 1º e o 3º. Exiba o ranking completo e a diferença.

jogador_1 = int(input('Digite a pontuação do 1º jogador: '))
jogador_2 = int(input('Digite a pontuação do 2º jogador: '))
jogador_3 = int(input('Digite a pontuação do 3º jogador: '))

if 0 <= jogador_1 <= 1000 and 0 <= jogador_2 <= 1000 and 0 <= jogador_3 <= 1000:
    if jogador_1 == jogador_2 == jogador_3:
        print('Todos os jogadores empataram com a mesma pontuação.')
    else:
        if jogador_1 >= jogador_2 and jogador_1 >= jogador_3:
            primeiro = jogador_1
            if jogador_2 >= jogador_3:
                segundo = jogador_2
                terceiro = jogador_3
            else:
                segundo = jogador_3
                terceiro = jogador_2
        elif jogador_2 >= jogador_1 and jogador_2 >= jogador_3:
            primeiro = jogador_2
            if jogador_1 >= jogador_3:
                segundo = jogador_1
                terceiro = jogador_3
            else:
                segundo = jogador_3
                terceiro = jogador_1
        else:
            primeiro = jogador_3
            if jogador_1 >= jogador_2:
                segundo = jogador_1
                terceiro = jogador_2
            else:
                segundo = jogador_2
                terceiro = jogador_1

        diferenca = primeiro - terceiro

        print(f'\nRANKING FINAL:\n1º lugar: {primeiro}\n2º lugar: {segundo}\n3º lugar: {terceiro}\nDiferença entre 1º e 3º: {diferenca}')
else:
    print('Pontuação inválida. Digite valores entre 0 e 1000.')
```

```python
# 24 - Calculadora de Reajuste Salarial
#Leia o salário atual e o percentual de avaliação de desempenho (0 a 100). Valide ambos.
#Use elif para o reajuste: avaliação ≥ 90 → 15%, ≥ 75 → 10%, ≥ 60 → 5%, abaixo → 0%.
#Com if aninhado, verifique se o novo salário ultrapassa o teto de R$15.000 — se sim, ajuste ao teto e exiba aviso. Exiba salário anterior, percentual aplicado, reajuste em reais e novo salário.

salario = float(input("Digite o salario atual: "))
desempenho = int(input("Digite a avaliação de desempenho: "))

if salario >= 0 and 0 <= desempenho <= 100:
    if desempenho >= 90:
        reajuste = salario * 1.15
        percentual = 15
    elif desempenho >= 75:
        reajuste = salario * 1.10
        percentual = 10
    elif desempenho >= 60:
        reajuste = salario * 1.05
        percentual = 5
    else:
        reajuste = salario
        percentual = 0
    if reajuste > 15000:
        print(f'O salário atingiu o teto máximo de R$15.000,00 - Salário anterior: R${salario:.2f} - Percentual aplicado: {percentual}% - Reajuste: R${15000 - salario:.2f} - Novo salário: R$15000.00')
    else:
        print(f'Salário anterior: R${salario:.2f} -Percentual aplicado: {percentual}% - Reajuste: R${reajuste - salario:.2f} - Novo salário: R${reajuste:.2f}')
else:
    print('Valores inválidos.')
```

```python
# 25 - Classificador de Triângulo com Ângulos
#Leia os três ângulos internos de um triângulo em graus. Valide que todos são positivos.
#Verifique se a soma é exatamente 180° — se não for, exiba "Triângulo inválido" e o valor da soma encontrada.
#Se válido, use elif para classificar quanto aos ângulos: acutângulo (todos < 90°), retângulo (algum == 90°) ou obtusângulo (algum > 90°).
#Com if aninhado, classifique também quanto aos lados com base nos ângulos iguais.

angulo1 = int(input('Digite o 1º angulo do triângulo em graus: '))
angulo2 = int(input('Digite o 2º angulo do triângulo em graus: '))
angulo3 = int(input('Digite o 3º angulo do triângulo em graus: '))

if angulo1 >= 0 and angulo2 >= 0 and angulo3 >= 0:
    if angulo1 + angulo2 + angulo3 != 180:
        print(f"Trinângulo inválido. Soma dos angulos = {angulo1+angulo2+angulo3}°")
    elif angulo1 < 90 and angulo2 < 90 and angulo3 < 90:
        print(f"Trinângulo acutângulo (todos os angulos < 90°)")
        if angulo1 == angulo2 == angulo3:
            print("Classificação quanto aos lados: Equilátero")
        elif angulo1 == angulo2 or angulo1 == angulo3 or angulo2 == angulo3:
            print("Classificação quanto aos lados: Isósceles")
        else:
            print("Classificação quanto aos lados: Escaleno")
    elif angulo1 == 90 or angulo2 == 90 or angulo3 == 90:
        print(f"Trinângulo retângulo (um angulo é = 90°)")
        if angulo1 == angulo2 == angulo3:
            print("Classificação quanto aos lados: Equilátero")
        elif angulo1 == angulo2 or angulo1 == angulo3 or angulo2 == angulo3:
            print("Classificação quanto aos lados:  Isósceles")
        else:
            print("Classificação quanto aos lados: Escaleno")
    else:
        print(f"Trinângulo obtusângulo (um angulo é > 90°)")
        if angulo1 == angulo2 == angulo3:
            print("Classificação quanto aos lados: Equilátero")
        elif angulo1 == angulo2 or angulo1 == angulo3 or angulo2 == angulo3:
            print("Classificação quanto aos lados: Isósceles")
        else:
            print("Classificação quanto aos lados: Escaleno")
```

```python
# 27 - Verificador de Senha com Nível de Segurança
#Leia uma senha e uma confirmação. Na camada externa, verifique se as duas são iguais — se não, exiba "Senhas não conferem" e encerre a lógica.
#Se iguais, use elif para classificar: menos de 6 caracteres → "Fraca", 6-9 → "Média", 10 ou mais → "Forte".
#Com if aninhado, verifique se a senha é composta apenas por dígitos numéricos (use a condição senha.isdigit()) — se sim, rebaixe um nível. Exiba o nível final.

senha = input('Digite a senha: ')
confirmacao = input('Digite novamente a senha: ')

if senha != confirmacao:
    print('Senhas não conferem')
else:
    if 0 <= len(senha) < 6:
        nivel = 'Fraca'
    elif 6 <= len(senha) <= 9:
        nivel = 'Média'
    else:
        nivel = 'Forte'
    if senha.isdigit():
        if nivel == 'Forte':
            nivel = 'Média'
        elif nivel == 'Média':
            nivel = 'Fraca'
    print(nivel)
```

```python
# 28 - Simulador de Reserva de Assento
#Leia o número do assento (1 a 100) e a classe desejada usando match/case ("economica", "executiva", "primeira").
#Valide o número. Com if aninhado, verifique se a faixa do assento corresponde à classe: 1-10 → primeira, 11-30 → executiva, 31-100 → econômica.
#Se incompatível, exiba mensagem. Se compatível, calcule o preço (econômica R$350, executiva R$890, primeira R$1.500) e exiba a confirmação.

assento = int(input('Digite o número do assento: '))
classe = input('Digite a  classe (economica, executiva ou primeira): ').lower()

if 1 <= assento <= 100:
    match classe:
        case "primeira":
            if assento <= 10:
                print('Número do assento e classe compatíveis. Valor: R$1.500,00. Deseja reservar?')
            else:
                print("O número do assento não é compatível com a classe.")
        case "executiva":
            if 11 <= assento <= 30:
                print('Número do assento e classe compatíveis. Valor: R$890,00. Deseja reservar?')
            else:
                print("O número do assento não é compatível com a classe.")
        case "economica":
            if 31 <= assento <= 100:
                print('Número do assento e classe compatíveis. Valor: R$350,00. Deseja reservar?')
            else:
                print("O número do assento não é compatível com a classe.")
        case _:
            print("Classe inválida.")
else:
    print("Número de assento inválido.")
```

```python
# 29 - Simulador de Batalha por Tipos
#eia o tipo do atacante e o tipo do defensor usando match/case. Tipos: "fogo", "agua", "planta", "eletrico", "terra".
#Relações de vantagem: fogo vence planta, água vence fogo, planta vence água, elétrico vence água, terra vence elétrico.
#Com if aninhado, determine se é vantagem (dano x2), desvantagem (dano x0.5) ou neutro (dano x1). Leia o poder base do ataque, calcule o dano final e exiba o resultado.

atacante = input('Digite o tipo do atacante (fogo, agua, planta, eletrico, terra): ').lower()
defensor = input('Digite o tipo do defensor (fogo, agua, planta, eletrico, terra): ').lower()
poder = float(input('Digite o poder base do ataque: '))

match atacante:
    case 'fogo':
        if defensor == 'planta':
            dano_final = poder * 2
        elif defensor == 'agua':
            dano_final = poder * 0.5
        else:
            dano_final = poder
    case 'agua':
        if defensor == 'fogo':
            dano_final = poder * 2
        elif defensor == 'planta':
            dano_final = poder * 0.5
        else:
            dano_final = poder
    case 'planta':
        if defensor == 'agua':
            dano_final = poder * 2
        elif defensor == 'fogo':
            dano_final = poder * 0.5
        else:
            dano_final = poder
    case 'eletrico':
        if defensor == 'agua':
            dano_final = poder * 2
        elif defensor == 'terra':
            dano_final = poder * 0.5
        else:
            dano_final = poder
    case 'terra':
        if defensor == 'eletrico':
            dano_final = poder * 2
        elif defensor == 'agua':
            dano_final = poder * 0.5
        else:
            dano_final = poder
    case _:
        print('Tipo inválido.')

print(f'Dano final causado: {dano_final:.2f}')
```

```python
# 30 - Sistema Completo de Reserva de Hotel
#Leia o tipo de acomodação usando match/case ("standard", "luxo", "suite"), o número de diárias (validar como positivo) e se o hóspede tem cartão fidelidade (s/n).
#Defina a diária base de cada tipo como variável. Use elif aninhado para desconto por diárias: 1-3 → 0%, 4-7 → 10%, acima de 7 → 20%. Com if aninhado adicional, aplique mais 5% para cartão fidelidade.
#Valide que o total não seja inferior a R$100. Use variáveis descritivas (diaria_base, desconto_qtd, desconto_fidelidade, total_final) e exiba o resumo completo com f-string.

tipo = input('Digite o tipo de acomodação (standard, luxo, suite): ').lower()
diaria = int(input('Digite o número de diárias: '))
fidelidade = input('Possui cartão fidelidade(s/n)? ').lower()

standard = 280.00
suite = 450.00
luxo = 800.00

if diaria > 0:
    match tipo:
        case 'standard':
            diaria_base = standard * diaria
        case 'luxo':
            diaria_base = luxo * diaria
        case 'suite':
            diaria_base = suite * diaria
        case _:
            print('Tipo de acomodação inválido.')
    if 1 <= diaria <= 3:
        desconto_qtd = 0
    elif 4 <= diaria <= 7:
        desconto_qtd = diaria_base * 0.10
    else:
        desconto_qtd = diaria_base * 0.20

    subtotal = diaria_base - desconto_qtd

    if fidelidade == 's':
        desconto_fidelidade = subtotal * 0.05
    else:
        desconto_fidelidade = 0

    total_final = subtotal - desconto_fidelidade

    if total_final < 100:
        total_final = 100

    print(f'Resumo da reserva:\nTipo: {tipo}\nDiárias: {diaria}\nValor base: R${diaria_base:.2f}\nDesconto por quantidade: R${desconto_qtd:.2f}\nDesconto fidelidade: R${desconto_fidelidade:.2f}\nTotal final: R${total_final:.2f}')
else:
    print('Número de diárias inválido.')
```
