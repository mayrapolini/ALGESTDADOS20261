```python
# 02 - Escreva um programa que leia 10 nomes e 10 idades. Armazene os nomes em uma lista e as idades em outra. Em seguida, exiba o nome da pessoa mais velha.

nomes = []
idades = []

for i in range(4):
    nome = input(f'Digite o {i+1}º nome: ').capitalize()
    idade = input(f'Digite a {i+1}ª idade: ')
    nomes.append(nome)
    idades.append(idade)

maior_idade = max(idades)

indice = idades.index(maior_idade)

print(f"A pessoa mais velha é {nomes[indice]} com {maior_idade} anos.")
```

```python
# 03-Faça um programa que leia uma lista de 10 números e divida todos os valores ímpares por 3 e os pares por 2, substituindo os valores originais.
# (Use estruturas condicionais.)

numeros = []

for i in range(10):
    numero = float(input(f'Digite o {i+1}º número: '))
    if numero % 2 == 0:
        numero = numero / 2
    else:
        numero = numero / 3
    numeros.append(numero)

lista_formatada = []

for numero in numeros:
    lista_formatada.append(f'{numero:.2f}')

print(lista_formatada)
```

```python
# 04-Solicite ao usuário uma frase. Armazene em uma lista cada palavra que começa com uma vogal. Exiba a lista resultante.

frase = input('Digite uma frase: ')

vogais = []
palavra = ''

for letra in frase + ' ':

    if letra != ' ':
        palavra += letra
    else:
        if palavra != '' and palavra[0].lower() in 'aeiou':
            vogais.append(palavra)
        palavra = ''

if len(vogais) > 0:
    print(vogais)
else:
    print('Não há palavras que começam com vogais nesta frase.')
```

```python
# 05-Leia 10 números inteiros e armazene-os em uma lista. Em seguida, crie uma segunda lista apenas com os números primos da primeira.

numeros = []

for i in range(10):
    numero = int(input(f'Digite o {i+1}º número: '))
    numeros.append(numero)

primos = []
for numero in numeros:
    if numero > 1:
        divisores = 0

        for i in range(1,numero+1):
            if numero % i == 0:
                divisores += 1

        if divisores == 2:
            primos.append(numero)

if len(primos) > 0:
    print(primos)
else:
    print('Não foram localizados números primos.')
```

```python
# 06-Escreva um programa que leia uma lista de 8 elementos e verifique (estrutura condicional) se ela é palíndroma (lê-se igual de trás para frente).

itens = []

for i in range(8):
    item = input(f'Digite o {i+1}º item: ')
    itens.append(item)

for i in range(4):
    if itens[i] != itens[7-i]:
        print('Não é palíndromo')
        break
else:
    print('É palíndromo')
```

```python
# 12 - Gere uma matriz 5×5 com números aleatórios entre 1 e 99. Calcule a média dos elementos de cada linha e armazene os resultados em uma nova lista.

matriz = []


for i in range(5):
    linha = []

    for j in range(5):
        numero = int(input(f'Digite o valor [{i}][{j}]: '))
        if 1 <= numero < 99:
            linha.append(numero)
        else:
            print('Número inválido. Digite um número entre 1 e 99.')

    matriz.append(linha)

medias = []

for linha in matriz:
    soma = 0

    for numero in linha:
        soma += numero

    media = soma / len(linha)

for linha in matriz:
    print(linha)
print(medias)
```

```python
# 16 - Crie uma matriz 4×4 com números fornecidos pelo usuário. Encontre o menor valor de cada coluna e armazene em uma lista separada.

matriz = []

for i in range(4):
    linha = []

    for j in range(4):
        numero = int(input(f'Digite o valor [{i}][{j}]: '))
        linha.append(numero)

    matriz.append(linha)

print('Matriz:')
for linha in matriz:
    print(linha)

menor_valor = []

for j in range(4):
    menor = matriz[0][j]
    for i in range(4):
        if matriz[i][j] < menor:
            menor = matriz[i][j]

    menor_valor.append(menor)

print(f'menores valores: {menor_valor}')
```

```python
#17 - Dada uma matriz 3×3 com valores reais, calcule o determinante (sem usar bibliotecas externas). Use apenas listas e estruturas de repetição.

matriz = [[1,2,3],
          [4,5,6],
          [2,4,6],]

sarrus = []

for i in range(3):
    linha = []

    for j in range(3):
        linha.append(matriz[i][j])

    linha.append(matriz[i][0])
    linha.append(matriz[i][1])

    sarrus.append(linha)

for linha in sarrus:
    print(linha)

principal = 0

for j in range(3):
    produto = 1

    for i in range(3):
        produto = produto * sarrus[i][j + i]

    principal += produto

secundaria = 0

for j in range(3):
    produto = 1

    for i in range(3):
        produto = produto * sarrus[2 - i][j + i]

    secundaria += produto

    determinante = principal - secundaria

print(determinante)
```

```python
#18 - Faça um programa que leia duas matrizes 3×3 e calcule a matriz soma e a matriz diferença, armazenando os resultados em novas listas.

matriz1 = []

for i in range(3):
    linha = []

    for j in range(3):
        numero = int(input(f'Digite o valor [{i}][{j}]: '))
        linha.append(numero)

    matriz1.append(linha)

print('Matriz 1:')
for linha in matriz1:
    print(linha)

matriz2 = []

for i in range(3):
    linha = []

    for j in range(3):
        numero = int(input(f'Digite o valor [{i}][{j}]: '))
        linha.append(numero)

    matriz2.append(linha)

print('Matriz 2:')
for linha in matriz2:
    print(linha)

matriz_soma = []

for i in range(3):
    linha = []

    for j in range(3):
        numero = matriz1[i][j] + matriz2[i][j]
        linha.append(numero)

    matriz_soma.append(linha)

print('Matriz Soma:')
for linha in matriz_soma:
    print(linha)

matriz_diferenca = []

for i in range(3):
    linha = []

    for j in range(3):
        numero = matriz1[i][j] - matriz2[i][j]
        linha.append(numero)

    matriz_diferenca.append(linha)

print('Matriz Diferença:')
for linha in matriz_diferenca:
    print(linha)
```

```python
# 26 - Faça um programa que armazene em um dicionário o nome de 5 alunos e uma lista com suas 3 notas. Depois, exiba o nome e a média de cada aluno.

diario = {}

for i in range(5):
    nome = input('Digite o nome do aluno: ').capitalize()
    notas = []
    for j in range (3):
        nota = float(input(f'Digite a {j+1}ª nota do aluno: '))
        notas.append(nota)
    diario[nome] = notas

print(diario)

for nome, notas in diario.items():
    soma = 0
    for nota in notas:
        soma += nota

    media = soma / len(notas)

    print(f'Nome: {nome} - Média: {media:.2f}')
```

```python
# 27 - Crie um dicionário que armazene palavras e seus sinônimos. Depois, permita que o usuário digite uma palavra e veja seus sinônimos, se existirem.
# (Use estrutura condicional.)

dicionario = {}

while True:
    palavra = input('Digite uma palavra: ').lower()
    if palavra == "sair":
        break
    sinonimos = []
    while True:
        sinonimo = input('Digite os sinônimos da palavra ou "fim" para encerrar: ').lower()
        if sinonimo == 'fim':
            break
        else:
            sinonimos.append(sinonimo)
    dicionario[palavra] = sinonimos

print("DICIONÁRIO DE SINÔNIMOS")
print(dicionario)

busca = input('Digite a palavra que busca: ').lower()

if busca in dicionario:
    print(f'Palavra: {busca} - Sinônimos: {dicionario[busca]}')
else:
    print(f'A palavra {busca} não existe neste dicionário.')
```

```python
# 29 - Crie um dicionário onde as chaves são nomes de produtos e os valores são listas com preços praticados em três lojas diferentes.
#Encontre o produto com maior variação de preço.

pesquisa_preco = {
    'sofá': [1600, 2000, 1800],
    'geladeira': [2500, 2000, 2200],
    'fogão': [600, 800, 750]
}

maior_variacao = 0
produto_variacao = ''

for produto, precos in pesquisa_preco.items():

    maior = precos[0]
    menor = precos[0]

    for preco in precos:

        if preco > maior:
            maior = preco

        if preco < menor:
            menor = preco

    variacao = maior - menor

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_variacao = produto

print(f'Produto com maior variação: {produto_variacao}\nMaior variação de preço: R$ {maior_variacao}')
```

```python
# 30 - Implemente um programa que leia uma frase e armazene em um dicionário o número de ocorrências de cada palavra (ignorando maiúsculas/minúsculas).
 #(Use estruturas de repetição.)

ocorrencias = {}
palavra = ''

frase = input("Digite uma frase: ").lower()

for caractere in frase:
    if caractere != ' ':
        palavra += caractere
    else:
        if palavra in ocorrencias:
            ocorrencias[palavra] += 1
        else:
            ocorrencias[palavra] = 1

        palavra = ''

if palavra != '':
    if palavra in ocorrencias:
        ocorrencias[palavra] += 1
    else:
        ocorrencias[palavra] = 1

print(ocorrencias)
```
