```python
# 01 - Levantamento Salarial

soma = 0
quantidade = 0
maior = 0
qtd_acima_3000 = 0

for i in range(10):
    salario = float(input('Digite o salário do funcionário: '))

    soma += salario
    quantidade += 1

    if salario > maior:
        maior = salario

    if salario > 3000:
        qtd_acima_3000 += 1

print(f'Média salarial da emprea: {media:.2f}\nMaior salário: R$ {maior:.2f}\nFuncionários que recebm acima de R$ 3000: {qtd_acima_3000}')
```

```python
# 02 - Desempenho da Turma

soma = 0
quantidade = 0
aprovados = 0
reprovados = 0

while True:
    nota = float(input('Digite a nota do aluno (-1 para sair): '))

    if nota == -1:
        break

    soma += nota
    quantidade += 1

    if nota >= 7:
        aprovados += 1
    else:
        reprovados += 1

if quantidade > 0:
    media = soma / quantidade
    print(f'Média da turma: {media:.2f}\nQuantidade de alunos aprovados: {aprovados}\nQauntidade de alunos reprovados: {reprovados}')
else:
    print('Nenhuma nota foi informada.')
```

```python
# 03-Sistema Bancário de Saques

saldo = float(input('Digite o saldo inicial da conta: '))

while True:
    saque = float(input('Digite o valor que deseja sacar (0 para sair): '))

    if saque == 0:
        break

    if saque < 0:
        print('Valor inválido')
        continue

    if saque <= saldo:
        saldo -= saque
        print(f'Saque realizado! Saldo disponível: R${saldo:.2f}')
    else:
        print('Saldo insuficiente')
```

```python
# 04-Pesquisa Demográfica

soma = 0
quantidade = 0
maior = 0
maiores_18 = 0

while True:
    idade = int(input('Informe a idade do entrevistado (0 para sair): '))

    if idade == 0:
        break

    soma += idade
    quantidade += 1

    if idade > maior:
        maior = idade

    if idade >= 18:
        maiores_18 += 1

if quantidade > 0:
    media = soma / quantidade
    print(f'Média das idades registradas: {media:.2f}\nMaior idade registrada: {maior}\nQuantidade de pessoas entrevistadas maiores que 18 anos: {maior_18}')
else:
    print('Nenhuma idade foi informada.')
```

```python
# 05-Controle de Acesso

senha_correta = 2024

for i in range(3):
    senha_digitada = int(input('Digite a senha: '))

    if senha_digitada == senha_correta:
        print('Acesso liberado')
        break
    else:
        if i < 2:
            print(f'Senha incorreta. Você ainda tem {2 - i} tentativa(s)')
        else:
            print('Sistema bloqueado. Excedido o número de tentativas (3)')
```

```python
# 06-Vendas com Desconto

soma = 0

for i in range(8):
    valor = float(input('Digite o valor da venda: '))
    if valor > 100:
        valor = valor * 0.9
    soma += valor

print(f'Total da venda: R${soma:.2f} ')
```

```python
# 07-Análise de Sequência Numérica

pares = 0
impares = 0
soma = 0

while True:
    numero = int(input('Digite um número (0 para sair): '))

    if numero == 0:
        break

    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1

    if numero > 0:
        soma += numero

print(f'Quantidade de pares: {pares}\nQuantidade de ímpares: {impares}\nSoma dos números positivos: {soma}')
```

```python
# 08-Monitoramento de Peso

soma = 0
contador = 0
menor = float('inf')

for i in range(6):
    peso = float(input('Digite o peso em kg: '))

    soma += peso

    if peso > 80:
        contador += 1

    if peso < menor:
        menor = peso

media = soma / 6

print(f'Média de peso: {media:.2f}kg\nQuantidade de pessoas que possuem peso superior a 80kg: {contador}\nMenor peso registrado:{menor}')
```

```python
# 09-Votação Eletrônica

candidato_1 = 0
candidato_2 = 0
candidato_3 = 0

while True:
    voto = int(input('Digite seu voto (1, 2 ou 3, 0 para sair): '))

    if voto == 0:
        break

    if voto == 1:
        candidato_1 += 1
    elif voto == 2:
        candidato_2 += 1
    elif voto == 3:
        candidato_3 += 1
    else:
        print('Voto inválido')


if candidato_1 > candidato_2 and candidato_1 > candidato_3:
    vencedor = 1
elif candidato_2 > candidato_1 and candidato_2 > candidato_3:
    vencedor = 2
else:
    vencedor = 3

print(f'Candidato 1: {candidato_1} votos\nCandidato 2: {candidato_2} votos\nCandidato 3: {candidato_3} votos\nO candidato que obteve a maior quantidade de votos foi o candidato {vencedor}')
```

```python
# 10-Produção Semanal

soma = 0
contador = 0
maior = 0
menor_100 = 0

for i in range (7):
    quantidade = int(input('Digite a quantidade produzida hoje: '))
    soma += quantidade
    contador += 1

    if quantidade > maior:
        maior = quantidade

    if quantidade < 100:
        menor_100 += 1

media = soma / contador

print(f'Média da produção semanal: {media:.2f}\nMaior valor registrado: {maior}\nQuantidade de dias que a produção foi inferior a 100: {menor_100}')
```

```python
# 11-Múltiplos de Três

soma = 0
contador = 0

while True:
    numero = int(input('Digite um número inteiro e -1 para encerrar: '))

    if numero < -1:
        print('Digite apenas números inteiros ou -1 para encerrar.')
        continue

    if numero == -1:
        break

    if numero % 3 == 0:
        soma += numero
        contador += 1

if contador > 0:
    media = soma / contador

    print(f'Quantidade de números multiplos de 3: {contador}\nMédia: {media:.2f}')
else:
    print('Não foram registrados múltiplos de 3.')
```

```python
# 12-Altura dos Alunos

soma = 0
maior_que_media = 0

for i in range(5):
    altura = float(input('Digite a altura do aluno(m): '))
    soma += altura

media = soma / 5

for i in range(5):
    altura = float(input('Digite novamente a altura do aluno(m): '))

    if altura > media:
        maior_que_media += 1

print(f'Média geral: {media:.2f}m')

if maior_que_media > 0:
    print(f'Foram registrados {maior_que_media} alunos com altura acima da média.')
else:
    print('Não foram encontrados alunos com altura acima da média.')
```

```python
# 13 - Registro de Depósitos

soma = 0
contador = 0
maior = 0

while True:
    deposito = float(input('Digite o valor do depósito ou 0 para sair: '))

    if deposito == 0:
        break

    soma += deposito
    contador += 1

    if deposito > maior:
        maior = deposito

print(f'Total depositado: R${soma:.2f}\nMaior valor depositado: R${maior}\nQuantidade total de depósitos efetuados: {contador}')
```

```python
# 14-Verificador de Número Primo

numero = int(input('Digite um número inteiro: '))

divisores = 0

for i in range (1, numero+1):
    if numero % i == 0:
        divisores += 1

if divisores == 2:
    print(f'O número {numero} é primo\nDivisores: 2')
else:
    print(f'O número {numero} não é primo\nDivisores: {divisores}')
```

```python
# 15-Loja de Eletrônicos

soma = 0

while True:
    preco = float(input('Digite o valor do produto ou 0 para encerrar: '))

    if preco == 0:
        break

    if preco > 200:
        desconto = preco * 0.95
        soma += desconto
    else:
        desconto = preco
        soma += desconto

print(f'Valor total: R${soma}')
```

```python
# 16-Tempo de Atendimento

soma = 0
contador = 0

while True:
    tempo = int(input('Digite o tempo de atendimento (minutos) ou 0 para encerrar: '))

    if tempo == 0:
        break

    soma += tempo
    contador += 1

if contador > 0:
    media = soma / contador
    print(f'Total de clientes atendidos: {contador}\nTempo médio de atendimento (min): {media:.2f}')
else:
    print('Nenhum atendimento registrado.')
```

```python
# 17-Análise de 10 Números

contador_negativos = 0
soma_positivos = 0

maior = None

for i in range(10):
    numero = int(input('Digite um número inteiro: '))

    if numero < 0:
        contador_negativos += 1
    else:
        soma_positivos += numero

    if maior is None or numero > maior:
        maior = numero

print(f'Quantidade de números negativos: {contador_negativos}\nSoma dos números positivos: {soma_positivos}\nMaior número informado: {maior}')
```

```python
# 18-Consumo de Energia

soma = 0
preco_kwh = 5

for i in range(6):
    consumo = float(input('Digite o consumo (kWh): '))

    if consumo > 100:
        valor = consumo * preco_kwh * 0.95
    else:
        valor = consumo * preco_kwh

    soma += valor

print(f'Total arrecadado: R$ {soma:.2f}')
```

```python
# 19-Adivinhe o Número Secreto

segredo = 50
tentativa = 0

while True:
    numero = int(input('Tente adivinhar o número secreto: '))
    tentativa += 1

    if numero == segredo:
        print(f'Acertou! Foram realizadas {tentativa} tentativas')
        break
    elif numero > segredo:
        print(f'O número {numero} é maior que o secreto')
    else:
        print(f'O número {numero} é menor que o secreto')
```

```python
# 21-Pesquisa Eleitoral

candidato_A = 0
candidato_B = 0
candidato_C = 0
soma_validos = 0
soma_invalidos = 0

while True:
    voto = int(input('Digite o número do candidato que deseja votar (1, 2, 3 ou 0 para sair): '))

    if voto == 0:
        break

    if voto == 1:
        candidato_A += 1
        soma_validos += 1
    elif voto == 2:
        candidato_B += 1
        soma_validos += 1
    elif voto == 3:
        candidato_C += 1
        soma_validos += 1
    else:
        soma_invalidos += 1

if candidato_A > candidato_B and candidato_A > candidato_C:
    vencedor = 'candidato A'
elif candidato_B > candidato_A and candidato_B > candidato_C:
    vencedor = 'candidato B'
else:
    vencedor = 'candidato C'

print(f'Votos válidos: {soma_validos}\nVotos inválidos: {soma_invalidos}\nVencedor: {vencedor}')
```

```python
# 22-Estacionamento

soma = 0
taxa = 5

while True:
    tempo = int(input('Digite o tempo de permanência (horas) ou zero para parar: '))

    if tempo == 0:
        break

    if tempo > 5:
       valor = (taxa*0.95) * tempo
    else:
       valor = taxa * tempo

    soma += valor

print(f'Total diário arrecadado: R${soma:.2f}')
```

```python
#  23-Divisíveis por 2 e por 3

soma = 0
contador = 0
contador_todos = 0


while True:
    numero = int(input('Digite um número ou zero para parar: '))

    if numero == 0:
        break

    contador_todos += 1
    soma+= numero

    if numero % 2 == 0 and numero % 3 == 0:
        contador += 1

media = soma / contador_todos

print(f'Números são divisíveis simultaneamente por 2 e por 3: {contador}\nMédia de todos os números digitados: {media:.2f}')
```

```python
# 24-Classificação Escolar

quantidade_reprovado = 0
quantidade_recuperacao = 0
quantidade_aprovado = 0

for i in range(8):
    nota = float(input('Digite a nota do aluno: '))

    if 0 <= nota < 6:
        quantidade_reprovado += 1
    elif 6 <= nota < 7:
        quantidade_recuperacao += 1
    else:
        quantidade_aprovado += 1

print(f'Aprovado: {quantidade_aprovado}\nRecuperação: {quantidade_recuperacao}\nReprovado: {quantidade_reprovado}')
```

```python
# 25-Análise de Vendas

soma = 0
contador = 0

while True:
    valor = float(input('Digite o valor da venda ou 0 para sair: '))

    if valor == 0:
        break

    soma += valor
    contador += 1

if contador == 0:
    print('Nenhuma venda registrada.')
else:
    media = soma / contador

    quantidade = 0
    i = 0

    while i < contador:
        valor = float(input('Digite novamente os valores para comparação: '))

        if valor > media:
            quantidade += 1

        i += 1

    print(f'Média de vendas: R$ {media:.2f}\nQuantidade acima da média: {quantidade}')
```

```python
# 27-Par ou Ímpar até N

N = int(input('Digite um número inteiro: '))

for i in range (1,N+1):
    if i % 2 == 0:
        print(f'O número {i} é par')
    else:
        print(f'O número {i} é ímpar')
```

```python
# 28-Caça aos Primos

soma = 0
primos = 0

while True:
    numero = int(input('Digite um número inteiro ou -1 para sair: '))

    if numero == -1:
        break

    if numero > 1:
        divisores = 0

        for i in range(1, numero + 1):
            if numero % i == 0:
                divisores += 1

        if divisores == 2:
            primos += 1
        else:
            soma += numero
    else:
        soma += numero

print(f'Quantidade de números primos: {primos}\nSoma dos números que não são primos: {soma}')
```

```python
# 29-Avaliação de Notas

soma = 0
contador = 0

while True:
    nota = float(input('Digite a nota ou 0 para encerrar: '))

    if nota == 0:
        break

    soma += nota
    contador += 1

if contador == 0:
    print('Nenhuma nota informada.')
else:
    media = soma / contador

    acima_media = 0
    i = 0

    while i < contador:
        nota = float(input('Digite novamente as notas: '))

        if nota > media:
            acima_media += 1

        i += 1

    print(f'Média geral: {media:.2f}\nNotas acima da média: {acima_media}')
```
